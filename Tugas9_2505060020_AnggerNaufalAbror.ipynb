**B. Kueri Bertingkat (Read)** *(bobot 20%)*

Tampilkan hasil dari **tiga** kueri berikut :
1. Seluruh menu kategori "Kopi" dengan `tingkat_kepahitan` di atas 3.
2. Seluruh menu dengan `harga` di antara 15000 dan 30000.
3. Menu yang **tidak memiliki** field `pedas_level` sama sekali.

```python
print("--- 1. Menu Kopi dengan tingkat kepahitan > 3 ---")
kueri_1 = koleksi.find({"kategori": "Kopi", "tingkat_kepahitan": {"$gt": 3}})
for menu in kueri_1:
    print(f"- {menu['nama_menu']} (Kepahitan: {menu['tingkat_kepahitan']}, Harga: Rp{menu['harga']:,})")

print("\n--- 2. Menu dengan harga antara Rp15.000 - Rp30.000 ---")
kueri_2 = koleksi.find({"harga": {"$gte": 15000, "$lte": 30000}})
for menu in kueri_2:
    print(f"- {menu['nama_menu']} (Kategori: {menu['kategori']}, Harga: Rp{menu['harga']:,})")

print("\n--- 3. Menu yang tidak memiliki field 'pedas_level' ---")
kueri_3 = koleksi.find({"pedas_level": {"$exists": False}})
for menu in kueri_3:
    print(f"- {menu['nama_menu']} (Kategori: {menu['kategori']})")
```

**D. Integrasi ke Spark** *(bobot 25%)*

Ambil seluruh data dari collection `menu`, ubah menjadi Spark DataFrame (ikuti pola Sub-bab 9.6 — perhatikan field yang mungkin kosong/`null` pada sebagian dokumen, seperti `tingkat_kepahitan` yang hanya dimiliki menu Kopi). Tampilkan:
1. `printSchema()` hasil konversi.
2. Rata-rata `harga` per `kategori` menggunakan `groupBy`.

```python
import pandas as pd
from pyspark.sql import SparkSession
from pyspark.sql.functions import avg, count, round

spark = SparkSession.builder \
    .appName("Tugas9-KopiNusantara") \
    .master("local[*]") \
    .getOrCreate()
spark.sparkContext.setLogLevel("ERROR")

dokumen_mongo = list(koleksi.find({}))

df_pandas = pd.DataFrame(dokumen_mongo)

df_pandas["_id"] = df_pandas["_id"].astype(str)

if "rekomendasi_barista" in df_pandas.columns:
    df_pandas["rekomendasi_barista"] = df_pandas["rekomendasi_barista"].fillna(False)

df_pandas = df_pandas.astype(object).where(pd.notna(df_pandas), None)

df_spark_menu = spark.createDataFrame(df_pandas)

print("--- 1. printSchema() Hasil Konversi ---")
df_spark_menu.printSchema()

print("--- Tampilan Data Spark DataFrame ---")
df_spark_menu.show(truncate=False)

print("--- 2. Rata-rata Harga per Kategori ---")
df_spark_menu.groupBy("kategori").agg(
    round(avg("harga"), 2).alias("rata_rata_harga"),
    count("nama_menu").alias("jumlah_menu")
).show()

spark.stop()
print("SparkSession berhasil ditutup.")
```

**E. Kesimpulan** *(bobot 20%)*

Tulis pada markdown cell (**minimal 100 kata**): apa tantangan yang anda temui saat mengonversi data MongoDB (dengan field yang tidak seragam antar dokumen) menjadi Spark DataFrame (yang mengharuskan skema tabel seragam)? Bagaimana anda mengatasinya?

Dalam proses integrasi basis data NoSQL MongoDB ke Apache Spark DataFrame, tantangan utama yang dihadapi terletak pada perbedaan pemodelan data. MongoDB menerapkan skema fleksibel (schema-less), di mana setiap dokumen dalam collection menu menyimpan struktur field yang tidak seragam—seperti kategori Kopi yang memiliki atribut tingkat_kepahitan, kategori Makanan yang memiliki atribut pedas_level, serta field opsional seperti rekomendasi_barista. Di sisi lain, Apache Spark DataFrame dirancang untuk pemrosesan skala besar yang mengharuskan skema terstruktur kaku (strict schema), di mana seluruh baris wajib memiliki kolom yang sama.
Saat data heterogen dari MongoDB diekstrak menggunakan PyMongo dan dikonversi via Pandas sebelum dibentuk menjadi Spark DataFrame, field yang tidak dimiliki oleh dokumen tertentu secara otomatis diterjemahkan menjadi nilai kosong (NaN di Pandas atau null di Spark). Tantangan ini diatasi dengan melakukan penyesuaian tipe data awal (mengubah _id ObjectId menjadi string) dan memanfaatkan fitur inferensi skema gabungan (schema union). Spark DataFrame secara alami menyerap seluruh gabungan field tersebut dan memperlakukan atribut yang absen sebagai nilai null, sehingga proses agregasi seperti groupBy dan kalkulasi avg(harga) tetap berjalan valid tanpa merusak eksekusi pipeline data.
